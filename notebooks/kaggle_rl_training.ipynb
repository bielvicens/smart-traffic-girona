## 6. Summary of Results

```python
import pandas as pd
df = pd.read_csv('experiments/rl/controller_metrics.csv')
summary = df.groupby(['scenario', 'controller'])[['average_waiting_time_seconds', 'average_travel_time_seconds', 'average_queue_length', 'average_stops']].agg(['mean', 'std']).round(2)
print(summary)
summary.to_csv('experiments/rl/evaluation_summary.csv')
```
